# Preparation of annotation tool outputs

Here we create the folders and set the parameters for the notebooks analysis.

```python
import sys
import os
import requests
from pathlib import Path
from benchannot.io.download import (
download_file,
download_and_extract,
download_reference_genomes
)
from benchannot.paths import REPOSITORY_ROOT, project_paths
```

```python
ORIGIN_PATHS = project_paths("origin")
REPRODUCED_PATHS = project_paths("reproduced")

DATA_GENOME_EUKARYOTE = ORIGIN_PATHS.genome_eukaryote
DATA_GENOME_PROKARYOTE = ORIGIN_PATHS.genome_prokaryote

DATA_ORIGIN = ORIGIN_PATHS.dataset_root
DATA_REPRODUCED = REPRODUCED_PATHS.dataset_root

DATA_ORIGIN_EUKARYOTE = ORIGIN_PATHS.eukaryote_tools
DATA_ORIGIN_PROKARYOTE = ORIGIN_PATHS.prokaryote_tools

DATA_REPRODUCED_EUKARYOTE = REPRODUCED_PATHS.eukaryote_tools
DATA_REPRODUCED_PROKARYOTE = REPRODUCED_PATHS.prokaryote_tools
```

```python
# Annotation tools used for each organism group
eukaryote_subfolders = [
    "eggnog",
    "gffread",
    "interproscan",
    "kofamscan",
    "pannzer"
]

prokaryote_subfolders = [
    "bakta",
    "eggnog",
    "pgap",
    "prokka"
]

# Create required directories
for base_dir, subfolders in [
    (DATA_ORIGIN_EUKARYOTE, eukaryote_subfolders),
    (DATA_ORIGIN_PROKARYOTE, prokaryote_subfolders),
    (DATA_REPRODUCED_EUKARYOTE, eukaryote_subfolders),
    (DATA_REPRODUCED_PROKARYOTE, prokaryote_subfolders)
]:
    for folder in subfolders:
        folder_dir = base_dir / folder
        folder_dir.mkdir(parents=True, exist_ok=True)
        print(f"{folder_dir.relative_to(REPOSITORY_ROOT)} created.")
```
